# **🎯 Python Number Guessing Game — Fun, Simple & Addictive! 🏆**

```
import random

def play_game(best_score):
    secret_number = random.randint(1, 100)

    print("\n🎯 Welcome to the Number Guessing Game!")
    print("🤔 I'm thinking of a number between 1 and 100.")
    print("💡 Can you guess it?")
    print("👉 Type 'exit' anytime to quit the game.")
    print("⏳ You have 7 attempts!\n")

    max_attempts = 7
    attempts = 0

    while attempts < max_attempts:
        guess_input = input("🔢 Enter your guess: ")

        # Allow user to quit anytime
        if guess_input.lower() == "exit":
            print("\n👋 You exited the game. Goodbye!")
            return "exit", best_score  # return exit flag

        # Validate numeric input
        if not guess_input.isdigit():
            print("⚠️ Please enter a valid number (1–100) or type 'exit' to quit.\n")
            continue

        guess = int(guess_input)
        attempts += 1

        if guess < secret_number:
            print("⬇️ Too low! Try again.\n")
        elif guess > secret_number:
            print("⬆️ Too high! Try again.\n")
        else:
            print(f"\n🎉 Congratulations! You guessed the number {secret_number} in {attempts} attempts. 🎯")

            if best_score is None or attempts < best_score:
                best_score = attempts
                print(f"🏆 New Record! 🎖️ You set a best score with {best_score} attempts.\n")
            else:
                print(f"📊 Your best score so far is {best_score} attempts.\n")
            break
    else:
        print(f"\n❌ Out of attempts! The number was {secret_number}.")
        if best_score:
            print(f"📊 Your best score so far is {best_score} attempts.\n")

    return "done", best_score


# Game loop with replay option + best score
best_score = None
while True:
    status, best_score = play_game(best_score)

    # If player exited, stop immediately (don’t ask replay)
    if status == "exit":
        if best_score:
            print(f"🥇 Your final best score was {best_score} attempts. Well done! 👏")
        break

    play_again = input("🔁 Do you want to play again? (yes/no): ").lower()
    if play_again != "yes":
        print("\n👋 Thanks for playing! See you next time. ✨")
        if best_score:
            print(f"🥇 Your best score was {best_score} attempts. 🏆")
        break
```
